```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("globalAirQuality.csv")
df
```

```python
# find records where pm 2.5 > overall average 2.5

avg_pm25 = df["pm25"].mean()

filtered = df.query("pm25 > @avg_pm25")
filtered
```

```python
# find records where pm 10 is greater than its 90th percentile 

pm10 = df["pm10"].quantile(0.90)
print(pm10)
df.query("pm10 > @pm10")
```

```python
sorted_df = df.sort_values("pm10",ascending = False)
sorted_df.iloc[0:1800]
```

```python
#find records where pm 2.5 is greater than the average pm 2.5 of their respectively country

country_wise_average_pm25 = df.groupby("country")["pm25"].transform("mean")

country_wise_average_pm25

filtered_records = df.query("pm25 > @country_wise_average_pm25")
filtered_records
```

```python
# find records where pm 2.5 > citywise median and NO2 is greater than the countrywise median

citywise_pm25_median = df.groupby("city")["pm25"].transform("median")
citywise_no2_median = df.groupby("country")["no2"].transform("median")

df.query("pm25 > @citywise_pm25_median & no2 > @citywise_no2_median")
```
