```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Product": ["A", "B", "C", "D", "E", "F", "G"],
    "Sales": [500, 800, 300, 900, 700, 600, 400],
    "Returns": [20, 100, 10, 50, 20, 80, 10],
    "Rating": [4.5, 3.8, 4.9, 4.2, 4.7, 3.5, 4.6],
    "Price": [1000, 2000, 500, 3000, 1500, 1200, 800]
}

df = pd.DataFrame(data)
```

```python
df["Revenue"] = df["Sales"]*df["Price"]
df[["Product","Revenue"]]
```

```python
df["Return Percentage"] = round(df["Returns"]/df["Sales"]*100,2)
df[["Product","Return Percentage"]]
```

```python
filtered_products = df[(df["Rating"]>4.5) & (df["Return Percentage"]<5)]
filtered_products
```

```python
df["Product Score"] = df["Rating"]*20-df["Return Percentage"]*2
df[["Product","Product Score"]]
```

```python
df["Rank"] = df["Product Score"].rank(ascending = False,method = "dense")
df[["Product","Rank"]]
```

```python
best_performing_product = df.loc[df["Product Score"].idxmax()]
# print(best_performing_product)
best_performing_product[["Product","Product Score"]]

best_performing_product["Product Score"].item()
```
