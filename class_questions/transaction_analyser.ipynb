```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Transaction": [101, 102, 103, 104, 105, 106, 107],
    "Account": ["A1", "A2", "A1", "A3", "A2", "A1", "A3"],
    "Amount": [500, 20000, 1500, 50000, 300, 25000, 700],
    "Status": ["Success", "Success", "Failed", "Success", "Success", "Success", "Failed"]
}

df = pd.DataFrame(data)

df
```

```python
# print(df["Amount"])
transactions_success_above_10000 = df[(df["Amount"]>=10000) & (df["Status"]=="Success")]
print(transactions_success_above_10000)
```
