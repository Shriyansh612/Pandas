```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Claim_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Customer": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Claim_Amount": [5000, 45000, 12000, 80000, 15000, 95000, 25000, 60000],
    "Customer_Age": [22, 45, 30, 55, 25, 60, 40, 35],
    "Previous_Claims": [0, 4, 1, 6, 0, 8, 2, 5],
    "Policy_Years": [3, 10, 2, 15, 1, 20, 5, 8],
    "Status": ["Approved", "Pending", "Approved", "Rejected", "Pending", "Pending", "Approved", "Rejected"]
}

df = pd.DataFrame(data)
```

```python
avg_claim_amount = df["Claim_Amount"].mean()
print(avg_claim_amount)
```

```python
claims_above_avg = df[df["Claim_Amount"]>avg_claim_amount]
claims_above_avg[["Claim_ID","Claim_Amount"]]
```

```python
filtered_cust = df[df["Previous_Claims"]>3]
filtered_cust[["Claim_ID","Customer","Previous_Claims"]]
```

```python
total_approved_claim_amount = df[df["Status"]=="Approved"]["Claim_Amount"].sum()
total_approved_claim_amount.item()
```

```python
conditions = [(df["Claim_Amount"]>50000) & (df["Previous_Claims"]>3),df["Claim_Amount"]>50000]

categories = ["Detailed Review","Amount Review"]

df["Review_Category"] = pd.Series("Standard Review",index = df.index)

df.loc[conditions[0],"Review_Category"] = categories[0]
df.loc[(conditions[1]) & (~conditions[0]),"Review_Category"] = categories[1]
df[["Claim_ID","Customer","Review_Category"]]
```

```python
total_claims = df["Previous_Claims"].sum()

percentage_of_claims = df.groupby("Status")["Previous_Claims"].sum()/total_claims*100
percentage_of_claims
```

```python
status_per = df["Status"].value_counts(normalize=True)*100
status_per
```

```python
df["Rank"] = df["Claim_Amount"].rank(ascending = True,method = "dense").astype(int)
ranked_customers = df.sort_values(by = "Rank")
df[["Claim_ID","Customer","Rank"]]
ranked_customers[["Claim_ID","Customer","Rank"]]
```
