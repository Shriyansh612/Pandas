```python
data = {
    "Customer": ["A", "B", "C", "D", "E", "F", "G"],
    "Plan": ["Basic", "Premium", "Basic", "Premium", "Standard", "Basic", "Standard"],
    "Monthly_Fee": [199, 599, 199, 599, 399, 199, 399],
    "Months_Active": [2, 12, 1, 18, 6, 3, 15],
    "Cancelled": ["Yes", "No", "Yes", "No", "No", "Yes", "No"]
}

df = pd.DataFrame(data)
```

```python
df["Revenue"] = df["Monthly_Fee"]*df["Months_Active"]
df
```

```python
cancellation_percentage = np.sum(df["Cancelled"]=="Yes")/len(df)*100
print(f"{cancellation_percentage:.2f}%")
```

```python
result = df.groupby("Plan")["Cancelled"].apply(
    lambda x: str((x=="Yes").sum()/len(x)*100)+"%"
)
result
```

```python
highest_cancelled_plan = result.sort_index().head(1)
print(highest_cancelled_plan)
```

```python
average_active_months = df.groupby("Cancelled")["Months_Active"].mean()
print(average_active_months)
```
