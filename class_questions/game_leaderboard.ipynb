```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Player": ["A", "B", "C", "D", "E", "F", "G"],
    "Kills": [50, 70, 40, 90, 60, 80, 30],
    "Wins": [5, 7, 3, 9, 6, 8, 2],
    "Matches": [10, 10, 10, 10, 10, 10, 10],
    "Deaths": [20, 15, 25, 10, 18, 12, 30]
}

df = pd.DataFrame(data)
```

```python
df["Win Percentage"] = df["Wins"]/df["Matches"]*100
df
```

```python
df["K/D"] = df["Kills"]/df["Deaths"]
df
```

```python
df["Performance Score"] = df["Kills"]*0.4 + df["Wins"]*10 + df["Win Percentage"]*0.3
df
```

```python
df["Rank"] = df["Performance Score"].rank(
    ascending=False,
    method="dense"
).astype(int)
df
```

```python
top_three = df.sort_values(
    by = "Performance Score",
    ascending = False
).head(3)
print(top_three[["Player","Performance Score","Rank"]])
```

```python
print(df[(df["K/D"]>4) & (df["Win Percentage"]>70)][["Player","K/D","Win Percentage"]])
```
