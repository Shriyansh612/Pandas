```python
import numpy as np
import pandas as pd
```

```python
data = {
    "City": ["Agra", "Delhi", "Agra", "Noida", "Delhi", "Agra", "Noida", "Delhi"],
    "Business": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Revenue": [50000, 90000, 30000, 75000, 60000, 80000, 40000, 100000],
    "Employees": [5, 10, 3, 8, 6, 7, 4, 12]
}

df = pd.DataFrame(data)
```

```python
city_wise_revenue = df.groupby("City")["Revenue"].sum()
city_wise_revenue
```

```python
average_revenue_per_business = df.groupby("City")["Revenue"].mean()
average_revenue_per_business
```

```python
highest_revenue_city = city_wise_revenue.idxmax()
print(highest_revenue_city)
print(city_wise_revenue.argmax())
```

```python
top_business_in_each_city = df.loc[df.groupby("City")["Revenue"].idxmax()]
print(top_business_in_each_city[["City","Business","Revenue"]])
```

```python
num_of_emp_per_city = df.groupby("City")["Employees"].sum()
num_of_emp_per_city
```

```python
# print(city_wise_revenue,type(city_wise_revenue))
new_df = city_wise_revenue.reset_index()
# print(new_df,type(new_df))
filter_cities = new_df[new_df["Revenue"]>150000]
filter_cities
```

```python
new_data = df[df["City"].isin(filter_cities["City"])]
new_data
```
