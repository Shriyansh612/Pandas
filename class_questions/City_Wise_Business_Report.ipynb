```python
import numpy as np
import pandas as pd
```

```python
data = {
    "City": ["Agra", "Delhi", "Agra", "Noida", "Delhi", "Agra", "Noida", "Delhi"],
    "Business": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Revenue": [50000, 90000, 30000, 75000, 60000, 80000, 40000, 100000],
    "Employees": [5, 10, 3, 8, 6, 7, 4, 12]
}

df = pd.DataFrame(data)
```

```python
city_wise_revenue = df.groupby("City")["Revenue"].sum()
city_wise_revenue
```

```python
average_revenue_per_business = df.groupby("City")["Revenue"].mean()
average_revenue_per_business
```

```python
print(city_wise_revenue.sort_values(ascending=False).head(1))
print(city_wise_revenue.iloc[[city_wise_revenue.argmax()]])
print(city_wise_revenue.idxmax(),city_wise_revenue.max()) #best
```

```python
highest_business_in_each_city = df.groupby("City")["Revenue"].max()
highest_business_in_each_city = highest_business_in_each_city.to_frame(name = "Maximum Revenue").reset_index()
highest_business_in_each_city
```

```python
num_of_emp_per_city = df.groupby("City")["Employees"].sum()
num_of_emp_per_city
```

```python
new_df = df[city_wise_revenue>150000]
new_df
```
