```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Department": ["Cardiology", "Neurology", "Emergency", "Orthopedic", "ICU", "Pediatrics"],
    "Doctors": [12, 8, 15, 10, 6, 9],
    "Patients": [180, 140, 300, 160, 120, 100],
    "Beds": [40, 30, 50, 35, 20, 40],
    "Equipment": [35, 20, 45, 25, 18, 30]
}

df = pd.DataFrame(data)
```

Calculate the patients per doctor ratio

```python
df["Patients/Doc"] = df["Patients"]/df["Doctors"]
df[["Department","Patients/Doc"]]
```

Calculate the bed utilization ratio using Patients and Beds

```python
df["Bed Utilization Ratio"] = df["Patients"]/df["Beds"]
df[["Department","Bed Utilization Ratio"]]
```

Identify departments where patients per doctor exceed 20

```python
departments1 = df[df["Patients/Doc"]>=20]["Department"]
departments1
```

Find Departments where patients exceed available beds

```python
departments2 = df[df["Patients"]>df["Beds"]]["Department"]
departments2
```

Create a "Resource Status" column based on the following conditions: 

Patients per doctor above 20 --> High Pressure

Otherwise --> Normal

```python
df["Resource Status"] = df["Patients/Doc"].apply(
    lambda x: "High Pressure" if x>=20 else "Normal"
)
df[["Department","Resource Status"]]
```

Rank Departments based on patients per doctor

```python
df["Rank"] = df["Patients/Doc"].rank(ascending = False,method = "dense").astype(int)
ranked_data = df.sort_values(by = "Rank")
ranked_data[["Department","Rank"]]
```

Identify the department with the highest resource pressure

```python
# print(df)
highest_department = df.loc[df["Patients/Doc"].argmax()]
highest_department
```
