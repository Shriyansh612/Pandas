```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "SSD", "RAM", "Headphones"],
    "Opening_Stock": [50, 200, 150, 40, 100, 120, 80],
    "Purchased": [30, 100, 50, 20, 60, 40, 30],
    "Sold": [65, 220, 170, 45, 130, 100, 60],
    "Unit_Price": [55000, 500, 1200, 15000, 4500, 3000, 2000]
}

df = pd.DataFrame(data)
```

```python
df["Closing Stock"] = df["Opening_Stock"] +df["Purchased"] - df["Sold"]
df[["Product","Closing Stock"]]
```

```python
df["Total Sales"] = df["Sold"] * df["Unit_Price"]
df[["Product","Total Sales"]]
```

```python
df["Stock Status"] = df["Closing Stock"].apply(
    lambda x: "Out of Stock" if x==0 else "Low Stock" if x<=20 else "Available" 
)
df[["Product","Stock Status"]]
```

```python
filter_products = df[df["Sold"]>df["Opening_Stock"]]
filter_products["Product"]
```

```python
df["Inventory Value"] = df["Closing Stock"]*df["Unit_Price"]
print(df[["Product","Inventory Value"]])

total_inventory_value = (df["Closing Stock"]*df["Unit_Price"]).sum()
print(total_inventory_value)
```

```python
df["Rank"] = df["Total Sales"].rank(ascending = False,method = "dense")
df[["Product","Rank"]]
```

```python
products_to_be_restocked = df[(df["Stock Status"]=="Out of Stock") ^ (df["Stock Status"]=="Low Stock")]["Product"]
products_to_be_restocked
```
