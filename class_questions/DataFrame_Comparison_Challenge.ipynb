```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Name": ["Aman", "Riya", "Karan", "Neha"],
    "City": ["Agra", "Delhi", "Noida", "Agra"]
})

df2 = pd.DataFrame({
    "ID": [3, 4, 5, 6],
    "Name": ["Karan", "Neha", "Rohit", "Simran"],
    "City": ["Noida", "Agra", "Delhi", "Agra"]
})
```

```python
common_customers = pd.merge(df1,df2,on = "ID",how = "inner")
common_customers[["ID","Name_x","City_x"]]
```

```python
only_in_df1 = df1[~df1["ID"].isin(df2["ID"])]
only_in_df1
```

```python
only_in_df2 = df2[~df2["ID"].isin(df1["ID"])]
only_in_df2
```

```python
df = pd.concat([df1,df2],ignore_index=True)
df
```

```python
unique_data = df.drop_duplicates(subset = "ID",keep = "first")
unique_data
```

```python
final_data = unique_data.reset_index(drop=True)
final_data
```
