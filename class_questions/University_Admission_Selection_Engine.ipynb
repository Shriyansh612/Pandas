```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Entrance_Score": [92, 85, 78, 95, 88, 72, 90, 82],
    "Academic_Percentage": [85, 90, 75, 80, 95, 88, 70, 82],
    "Interview_Score": [80, 95, 85, 75, 90, 80, 88, 92],
    "Category": ["General", "OBC", "General", "SC", "OBC", "General", "SC", "OBC"]
}

df = pd.DataFrame(data)
```

1. Calculate a weighted score using:

· Entrance Score: 50%
· Academic Percentage: 30%
· Interview Score: 20%

```python
df["Weighted Score"] = df["Entrance_Score"]*0.5 + df["Academic_Percentage"]*0.3 + df["Interview_Score"]*0.2
df[["Student","Weighted Score"]]
```

2. Rank students by weighted score.

```python
df["Rank"] = df["Weighted Score"].rank(ascending = False,method = "dense").astype(int)
df[["Student","Rank"]]

sorted_df = df.sort_values(by = "Rank",ignore_index=True)
sorted_df[["Student","Rank"]]
```

3. Find students whose entrance score is above 85 and academic percentage is above 80.

```python
filter_students = df[(df["Entrance_Score"]>85 & (df["Academic_Percentage"]>80))]
print(filter_students["Student"])      #series
# print(filter_students[["Student"]])    #dataframe
```

4. Find the highest-scoring student in each category.

```python
highest_entrance = df.iloc[df["Entrance_Score"].argmax()][["Student","Entrance_Score"]]
print(highest_entrance)
highest_academic = df.iloc[df["Academic_Percentage"].argmax()][["Student","Academic_Percentage"]]
print(highest_academic)
highest_interview = df.iloc[df["Interview_Score"].argmax()][["Student","Interview_Score"]]
print(highest_interview)
```

```python
cols = ["Entrance_Score","Academic_Percentage","Interview_Score"]

for col in cols:
    print("Highest Student in",col)
    print(df.iloc[df[col].argmax()][["Student",col]],"\n")
```

5. Create an Admission_Status column:

. Weighted score above 85 - Selected

. Weighted score between 75 and 85 - Waiting List

. Otherwise - Rejected

```python
df["Admission_Status"] = df["Weighted Score"].apply(
    lambda x: "Selected" if x>=85 else "Waiting List" if 75<=x<85 else "Rejected"
)
df[["Student","Weighted Score","Admission_Status"]]
```
