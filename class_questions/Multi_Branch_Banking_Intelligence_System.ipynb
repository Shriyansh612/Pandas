```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Branch": ["Agra", "Delhi", "Noida", "Agra", "Delhi", "Noida", "Agra", "Delhi", "Noida"],
    "Account_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109],
    "Account_Type": ["Savings", "Current", "Savings", "Current", "Savings", "Current", "Savings", "Current", "Savings"],
    "Balance": [50000, 120000, 30000, 80000, 150000, 45000, 25000, 90000, 70000],
    "Transactions": [20, 35, 15, 40, 50, 25, 10, 30, 45],
    "Loans": [0, 50000, 10000, 30000, 0, 20000, 0, 60000, 15000]
}

df = pd.DataFrame(data)
```

1. Calculate the average account balance for every branch.

```python
avg_acc_balance_per_branch = df.groupby("Branch")["Balance"].mean()
avg_acc_balance_per_branch
```

2. Calculate the total loans issued by each branch.

```python
loans_per_branch = df.groupby("Branch")["Loans"].apply(
    lambda x: (x!=0).sum()
)
print(loans_per_branch)

loans_per_branch2 = (df["Loans"]!=0).groupby(df["Branch"]).sum()
print(loans_per_branch2)
```

3. Identify branches where average account balance exceeds ₹60,000.

```python
filtered_branches = avg_acc_balance_per_branch[avg_acc_balance_per_branch>60000]
filtered_branches
```

4. Find the account with the highest loan amount.

```python
highest_loan_account = df.loc[df["Loans"].argmax()]
highest_loan_account
```

5. Calculate the ratio of loans to total balance for each branch.

```python
loan_upon_balance_per_branch = df.groupby("Branch").apply(
    lambda x: x["Loans"].sum()/x["Balance"].sum()
)
loan_upon_balance_per_branch
```

```python
#method 2
grouped = df.groupby("Branch")[["Loans","Balance"]].sum()
loan_upon_balance_per_branch2 = grouped["Loans"]/grouped["Balance"]
print(grouped)
loan_upon_balance_per_branch2
```

6. Rank branches based on total account balance.

```python
total_acc_balance_per_branch = df.groupby("Branch")["Balance"].sum()
total_acc_balance_per_branch = total_acc_balance_per_branch.reset_index()
print(type(total_acc_balance_per_branch))
total_acc_balance_per_branch["Rank"] = total_acc_balance_per_branch["Balance"].rank(ascending = False,method = "dense")
total_acc_balance_per_branch
```

7. Create a Customer_Segment column:

· Balance above 100,000 - High Value

. Balance between 50,000 and 100,000 - Medium Value

. Otherwise - Low Value.

```python
df["Customer_Segment"] = df["Balance"].apply(
    lambda x: "High Value" if x>=100000 else "Medium Value" if 50000<=x<=100000 else "Low Value"
)
df[["Account_ID","Customer_Segment"]]
```

8. Identify the branch having the highest number of high-value customers.

```python
high_value_count = df.groupby("Branch").apply(
    lambda x: (x["Customer_Segment"]=="High Value").sum()
)
print(high_value_count)

highest_branch = high_value_count[[high_value_count.idxmax()]]
highest_branch
```
